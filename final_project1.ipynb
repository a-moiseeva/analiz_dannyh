# **а)Проверить на однородность по сумме баллов**

# ГТО и не ГТО

```python
df = pd.DataFrame(df, columns=['ГТО.оценка','Сумма.баллов'])
print(df)
GTOp = df[df['ГТО.оценка'] > 0]
GTOm = df[df['ГТО.оценка'] == 0]
print(GTOp)
print(GTOm)
print(ttest_ind(GTOm['Сумма.баллов'] , GTOp['Сумма.баллов'] ))  #критерий Стьюдента
print(sps.ks_2samp(GTOm['Сумма.баллов'] , GTOp['Сумма.баллов'])) #Критерий Колмогорова-Смирнова
```

Предполагаем, что баллы должны быть подченены нормальному распределению.
Используем критерий Колмогорова-Смирнова и критерий Стьюдента.

**Можно сказать, что однородно, т.к. pvalue >
5%.**

ГТО не влияет на результат по баллам.

#  **б) Проверить на независимость баллы по русскому и **математике****

```python
from scipy.stats import pearsonr
from scipy.stats import kendalltau
from scipy.stats import spearmanr
df = pd.read_csv('https://raw.githubusercontent.com/zzzf64/AD/main/data3.csv?token=GHSAT0AAAAAABTUVLBL762HOMOHMLUBQPPSYS4NZ3A', delimiter=',')
df = df[df['Оценка.4'] >=0]
df = df[df['Оценка.2'] >=0]
df = pd.DataFrame(df, columns=['Оценка.4','Оценка.2'])
print(df)
print('Пирсон, линейная зависимость',pearsonr(df['Оценка.4'],df['Оценка.2']))
print(kendalltau(df['Оценка.4'],df['Оценка.2']))
print(spearmanr(df['Оценка.4'],df['Оценка.2']))
```

Коэффиценты линейной и монотонной корреляции очень маленькие, значит, корреляции почти нет.

p-value для теста говорит, что нет корреляции, это наша гипотеза. Т.к. у нас p-value маленькое, значит принимаем альтернативу -- корреляция есть, но просто не линейная и не монотонная.
